```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
df = pd.read_csv("../data/processed/cleaned_supply_chain.csv")
```

```python
df.head()
```

```python
print("Rows and Columns:", df.shape)
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df.hist(figsize=(18, 15), bins=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Delivery Status')
plt.xticks(rotation=45)
plt.title("Delivery Status Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Late_delivery_risk')
plt.title("Late Delivery Risk")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Type')
plt.title("Shipping Type")
plt.show()
```

```python
plt.figure(figsize=(14,10))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    cmap="coolwarm",
    annot=False
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Sales per customer"], bins=30, kde=True)

plt.title("Sales per Customer Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Benefit per order"], bins=30, kde=True)

plt.title("Benefit per Order Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Sales per customer"])

plt.title("Sales per Customer Boxplot")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Days for shipping (real)"])

plt.title("Actual Shipping Days")
plt.show()
```

```python
plt.figure(figsize=(12,6))

top_categories = df["Category Name"].value_counts().head(10)

sns.barplot(
    x=top_categories.values,
    y=top_categories.index
)

plt.title("Top 10 Product Categories")
plt.xlabel("Number of Orders")
plt.ylabel("Category")
plt.show()
```

```python
plt.figure(figsize=(12,6))

top_cities = df["Customer City"].value_counts().head(10)

sns.barplot(
    x=top_cities.values,
    y=top_cities.index
)

plt.title("Top 10 Customer Cities")
plt.xlabel("Orders")
plt.ylabel("City")
plt.show()
```

```python
plt.figure(figsize=(14,6))

sales = df.groupby("Category Name")["Sales per customer"].mean().sort_values(ascending=False)

sns.barplot(
    x=sales.values,
    y=sales.index
)

plt.title("Average Sales per Customer by Category")
plt.show()
```

```python
plt.figure(figsize=(14,6))

risk = df.groupby("Category Name")["Late_delivery_risk"].mean().sort_values(ascending=False)

sns.barplot(
    x=risk.values,
    y=risk.index
)

plt.title("Average Late Delivery Risk by Category")
plt.show()
```

## Business Insights

- The dataset contains over 180,000 supply chain records.
- No duplicate records were found after cleaning.
- Most deliveries were completed on time.
- Some product categories generated higher average sales than others.
- Sales distribution is right-skewed with several high-value outliers.
- Certain product categories showed higher late delivery risk.
- Numerical features such as shipping days and delivery risk showed meaningful correlations.
