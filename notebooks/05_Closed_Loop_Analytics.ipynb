```python
import pandas as pd
from datetime import datetime
```

```python
executed_decision = {
    "Shipment_ID": 1001,
    "Predicted_Delay": 5,
    "Recommended_Action": "Secondary Supplier",
    "Predicted_Cost": 12500,
    "Execution_Date": datetime.now()
}

decision_df = pd.DataFrame([executed_decision])

decision_df
```

```python
decision_df.to_csv(
    "../data/processed/executed_decisions.csv",
    index=False
)

print("Decision saved successfully!")
```

```python
# Simulated actual outcome after executing the decision

actual_outcome = {
    "Shipment_ID": 1001,
    "Actual_Cost": 13800,
    "Actual_Delay": 6,
    "Outcome_Status": "Completed"
}

actual_df = pd.DataFrame([actual_outcome])

actual_df
```

```python
comparison = decision_df.merge(
    actual_df,
    on="Shipment_ID"
)

comparison["Cost_Difference"] = (
    comparison["Actual_Cost"] -
    comparison["Predicted_Cost"]
)

comparison["Delay_Difference"] = (
    comparison["Actual_Delay"] -
    comparison["Predicted_Delay"]
)

print("Decision Outcome")
print("=" * 40)

print(comparison[
    [
        "Shipment_ID",
        "Recommended_Action",
        "Predicted_Cost",
        "Actual_Cost",
        "Cost_Difference",
        "Predicted_Delay",
        "Actual_Delay",
        "Delay_Difference"
    ]
])
```

```python
# Calculate decision performance

comparison["Cost_Error_Percentage"] = (
    abs(comparison["Cost_Difference"]) /
    comparison["Predicted_Cost"]
) * 100

comparison["Delay_Error_Percentage"] = (
    abs(comparison["Delay_Difference"]) /
    comparison["Predicted_Delay"]
) * 100

# Define a simple decision performance score
comparison["Decision_Score"] = (
    100
    - comparison["Cost_Error_Percentage"]
    - comparison["Delay_Error_Percentage"]
)

print("Decision Performance")
print("=" * 40)

print(comparison[
    [
        "Shipment_ID",
        "Recommended_Action",
        "Cost_Error_Percentage",
        "Delay_Error_Percentage",
        "Decision_Score"
    ]
])
```

```python
# Save closed-loop feedback

feedback_columns = [
    "Shipment_ID",
    "Recommended_Action",
    "Predicted_Cost",
    "Actual_Cost",
    "Cost_Difference",
    "Predicted_Delay",
    "Actual_Delay",
    "Delay_Difference",
    "Cost_Error_Percentage",
    "Delay_Error_Percentage",
    "Decision_Score"
]

feedback_df = comparison[feedback_columns]

feedback_df.to_csv(
    "../data/processed/decision_feedback.csv",
    index=False
)

print("Decision feedback saved successfully!")
```
